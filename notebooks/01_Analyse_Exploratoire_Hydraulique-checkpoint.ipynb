# VibroFlow AI - Analyse Exploratoire du Système Hydraulique

Ce notebook présente l'analyse des données du système hydraulique (dataset0) pour la maintenance prédictive et la mesure de débit.

```python
import sys
import os
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Robust module path setup
def setup_sys_path():
    curr = os.getcwd()
    while curr != os.path.dirname(curr):
        if os.path.exists(os.path.join(curr, 'src')):
            if curr not in sys.path: sys.path.insert(0, curr)
            src = os.path.join(curr, 'src')
            if src not in sys.path: sys.path.insert(0, src)
            return curr
        curr = os.path.dirname(curr)
    return None

project_root = setup_sys_path()

from data.loader import HydraulicDataLoader
from data.features import FeatureExtractor
# Use project_root for data path
data_path = os.path.join(project_root, 'dataset0')
loader = HydraulicDataLoader(data_path)
loader = HydraulicDataLoader('../dataset0')
targets = loader.load_targets()
print(f"Nombre de cycles : {len(targets)}")
```

## 1. Visualisation des Capteurs

Visualisons les signaux de pression, vibration et débit pour un cycle donné.

```python
cycle_idx = 0
ps1 = loader.load_sensor('PS1')[cycle_idx]
vs1 = loader.load_sensor('VS1')[cycle_idx]
fs1 = loader.load_sensor('FS1')[cycle_idx]

fig, ax = plt.subplots(3, 1, figsize=(12, 10))

ax[0].plot(ps1, color='red')
ax[0].set_title('Pression (PS1) - 100 Hz')
ax[0].set_ylabel('bar')

ax[1].plot(vs1, color='blue')
ax[1].set_title('Vibration (VS1) - 1 Hz')
ax[1].set_ylabel('Amplitude')

ax[2].plot(fs1, color='green')
ax[2].set_title('Débit (FS1) - 10 Hz')
ax[2].set_ylabel('l/min')

plt.tight_layout()
plt.show()
```

## 2. Analyse des Cibles

Répartition des classes pour les différentes conditions de maintenance.

```python
plt.figure(figsize=(15, 5))
for i, col in enumerate(['cooler', 'valve', 'pump_leakage', 'accumulator']):
    plt.subplot(1, 4, i+1)
    sns.countplot(data=targets, x=col)
    plt.title(f'Répartition : {col}')

plt.tight_layout()
plt.show()
```

## 3. Extraction de Features et Corrélation

Calculons les caractéristiques statistiques pour la vibration (VS1) et corrélons-les avec le débit.

```python
extractor = FeatureExtractor(sampling_freq=1)
vibration_data = loader.get_vibration_data()
flow_data = loader.get_flow_data()[0]

features_list = []
for i in range(len(vibration_data)):
    f = extractor.extract_all(vibration_data[i], include_wavelet=False)
    f['avg_flow'] = np.mean(flow_data[i])
    features_list.append(f)

df_features = pd.DataFrame(features_list)

plt.figure(figsize=(10, 6))
sns.regplot(data=df_features, x='time_rms', y='avg_flow', scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('Corrélation Vibration (RMS) vs Débit Moyen')
plt.show()
```
